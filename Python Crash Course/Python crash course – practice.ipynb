### Values and variables
A **variable** is a name for a value. The line `takings = 250` stores the number 250 under the name `takings`; after that, wherever you write `takings`, Python uses 250.

R users: Python uses `=` where you would write `<-`. It means "store", not "is equal to" – we will meet the test for equality in the next section.

Arithmetic works as you would expect: `+`, `-`, `*` and `/`, with brackets to control the order. Two things are different from R:

- Powers use `**`, so `3 ** 2` is 9. Writing `3 ^ 2` does **not** give an error – it quietly gives 1, because `^` means something else in Python. If a number looks oddly small, check for a `^`.
- `/` always gives a decimal: `7 / 2` is `3.5`. `//` divides and throws away the remainder: `7 // 2` is `3`.

Run the cell below, then try changing the numbers.

````
takings = 250
costs = 180
profit = takings - costs
print(profit)
print(3 ** 2, 3 ^ 2)
print(7 / 2, 7 // 2)
````

### Lists
A **list** holds several values in order, inside square brackets: `week = [250, 310, 180, 400, 520]`. It is the nearest thing to R's `c(250, 310, ...)`.

You get one item out with its **position** in square brackets. The big difference from R: **Python counts from 0**. So `week[0]` is the first item (Monday), `week[1]` the second, and `week[4]` the fifth (Friday). Asking for `week[5]` gives an `IndexError`, because there is no sixth item. You can also count from the end: `week[-1]` is the last item.

Python has built-in functions that work on a whole list: `len()` (how many items), `sum()`, `min()` and `max()`. The average is `sum(week) / len(week)`.

````
week = [250, 310, 180, 400, 520]
print(week[0])        # Monday – the first item is at position 0
print(week[4])        # Friday
print(len(week), sum(week), min(week), max(week))
print(sum(week) / len(week))
````

### Changing a list
You can replace an item by assigning to its position: `week[2] = 200`.

To add an item to the end, use `week.append(610)`. The dot means "do this operation to this list": `.append()` belongs to lists, and it changes the list in place. This is how simulations record their history – Hiro's code does `history.append(balance)` every step.

`week.index(520)` tells you the position of the value 520. Combined with `max()`, `week.index(max(week))` gives the position of the best day (counting from 0, of course).

**Watch for:** `week = week.append(610)`. It looks reasonable, but `.append()` changes the list and hands back nothing (`None`), so this line throws your list away and leaves `week` as `None`. Just write `week.append(610)` on its own.

````
week = [250, 310, 180, 400, 520]
week.append(610)      # add Saturday
print(week)
print(len(week))
best = week.index(max(week))
print(f"Best day is at position {best}")
````

### Indentation comes first
In R, curly brackets `{ }` mark which lines belong inside a loop. Python uses **indentation** instead: the lines pushed in under a line ending in a colon `:` are the ones inside it. This is the single biggest change for R users, and it matters, because moving one line in or out changes what the code does – often **without any error**.

Run the two cells below. They are identical except that the last line is indented in the first and not in the second. The first prints a running total after every item; the second prints the total once, after the loop has finished.

````
week = [250, 310, 180, 400, 520]
total = 0
for amount in week:
    total += amount
    print(total)      # indented: inside the loop, runs 5 times
````

````
week = [250, 310, 180, 400, 520]
total = 0
for amount in week:
    total += amount
print(total)          # not indented: after the loop, runs once
````

Some practical points about indentation:

- Use **4 spaces** per level. Pressing **Tab** in Colab inserts spaces for you. (Colab may indent with 2 spaces by default; Hiro's code uses 4. Either works, as long as you are consistent. To match Hiro, go to *Tools → Settings → Editor* and set the indentation width to 4.)
- A line that ends in `:` must be followed by at least one indented line.
- Code pasted from an email, a PDF or a web page sometimes contains **tab characters**, which look like spaces but are not. Mixing them with spaces gives an `IndentationError` or `TabError`. The fix is to delete the indent on that line and press Tab again.

### Loops
`for amount in week:` runs the indented lines once for each item in the list, with `amount` holding the current item each time round.

`for i in range(5):` runs 5 times, with `i` taking the values 0, 1, 2, 3, 4. As with list positions, it starts at 0, and the end number is **not** included. `range(1, 6)` gives 1 to 5. Simulations often use `range` to mean "this many steps", e.g. `for month in range(360):`.

`total += amount` is shorthand for `total = total + amount`. To build up a total, set it to 0 **before** the loop and add to it **inside** the loop. (A common bug is putting `total = 0` inside the loop, which resets it every time.)

````
for i in range(5):
    print("step", i)

for i in range(1, 4):
    print("counting from 1:", i)
````

### Decisions
`if` runs some lines only when a condition is true. `elif` ("else if") adds another condition to try, and `else` catches everything left over. Each line ends with a colon, and the lines it controls are indented.

Comparisons: `>`, `<`, `>=`, `<=`, `!=` (not equal) and `==` (equal). Note the **double** equals sign for testing: `=` stores a value, `==` asks whether two values are the same. Writing `if total = 5:` is an error.

To combine conditions, Python uses the words `and`, `or` and `not` (R uses `&`, `|` and `!`).

````
week = [250, 310, 180, 400, 520]
for amount in week:
    if amount > 400:
        print(amount, "great day")
    elif amount > 250:
        print(amount, "good day")
    else:
        print(amount, "quiet day")
````

### Pattern 1 · Step and record
Now put it together. Start with £1,000 in a savings account that adds 0.5% interest each month. Each time round the loop, update the balance, then **record** it by appending it to a list called `history`. After the loop, plot the history.

This is exactly the shape of Hiro's mortgage model, and of his epidemic model: a value is updated step by step, and the list lets you see the whole trajectory rather than just where it ended up.

To draw the plot we use `matplotlib`, Python's standard plotting library. `import matplotlib.pyplot as plt` loads it under the short name `plt` (like `library()` in R), and then `plt.plot()`, `plt.xlabel()` and so on build up the figure.

````
import matplotlib.pyplot as plt

balance = 1000
history = []
for month in range(24):
    balance = balance * 1.005     # add 0.5% interest
    history.append(balance)       # record it

print(f"After 24 months: £{balance:.2f}")

plt.figure()
plt.plot(history)
plt.xlabel("Month")
plt.ylabel("Balance (£)")
plt.title("Savings at 0.5% a month")
plt.show()
````

**Watch for:** if `history.append(balance)` is not indented, it runs once after the loop, and your plot has a single point instead of 24. If a plot looks empty, check the indentation of the line that records.

### Writing a function
```python
def add_interest(balance, rate):
    return balance * (1 + rate)
```
`def` says "I'm defining a function"; `add_interest` is its name; `balance` and `rate` are its **arguments** – the inputs you give it; the line ends with a colon, and the indented **body** says what it does. `return` hands a result back.

R users: this is the same as `add_interest <- function(balance, rate) { balance * (1 + rate) }`, except that Python needs the word `return` – it does not hand back the last value automatically.

Running the cell that defines a function does not print anything; it just teaches Python the new name. You then **call** it, e.g. `add_interest(1000, 0.005)`. If you get `NameError: name 'add_interest' is not defined`, you have not run the defining cell yet.

An argument can have a **default** value, used when you leave it out: with `def add_interest(balance, rate=0.005):`, calling `add_interest(1000)` uses 0.5%.

````
def add_interest(balance, rate=0.005):
    return balance * (1 + rate)

print(add_interest(1000, 0.01))   # both arguments given
print(add_interest(1000))         # rate left out, so the default 0.005 is used
new_balance = add_interest(500)
print(new_balance)
````

### `print` versus `return`
These look similar but do different jobs. `print` shows a value on screen for you to read; `return` hands the value back to the code that called the function, so it can be stored and used. A function with no `return` hands back `None`.

In the cell below, `shout_double` prints its answer but does not return it, so `x` ends up as `None`. Simulations need `return`, because the result of one run gets added up, averaged or compared.

````
def shout_double(n):
    print(n * 2)        # shows it, but does not hand it back

def double(n):
    return n * 2        # hands it back

x = shout_double(5)
print("x is", x)
y = double(5)
print("y is", y)
````

### Functions and outside variables
A function can read a variable that was set outside it. That is convenient, but risky: if the outside variable changes, the function's answer silently changes too. Hiro warns about this twice in his shop example. The safer habit is to pass anything a function needs in as an argument.

````
rate = 0.005

def grow(balance):
    return balance * (1 + rate)    # reads the outside variable rate

print(grow(1000))
rate = 0.05                        # someone changes rate further down the notebook…
print(grow(1000))                  # …and the same call now gives a different answer
````

### Randomness
Simulations use random numbers to stand in for things we cannot predict: whether a customer turns up, whether a coin lands heads. Python's `random` module provides them once you `import random`:

- `random.random()` gives a decimal between 0 and 1 (never exactly 1). `random.random() < 0.3` is true 30% of the time, which is how you make something happen with a given probability.
- `random.randint(1, 6)` gives a whole number from 1 to 6 – here **both** ends are included, unlike `range`.
- `random.choice(week)` picks a random item from a list.
- `random.seed(1)` fixes the starting point, so the "random" numbers come out the same every time you run the cell.

Run the cell below a few times. Then delete the `random.seed(1)` line and run it a few more times.

````
import random
random.seed(1)
print(random.random())
print(random.randint(1, 6))
print(random.choice(["heads", "tails"]))
````

### Pattern 2 · Wrap and repeat

Here, `flip_coins(n)` flips `n` coins and returns the number of heads. We then run 1,000 rounds of 10 flips and count how often we got at least 7 heads. That count divided by 1,000 estimates the probability.

````
import random

def flip_coins(n):
    heads = 0
    for i in range(n):
        if random.random() < 0.5:
            heads += 1
    return heads
````

````
random.seed(1)
count = 0
for run in range(1000):
    if flip_coins(10) >= 7:
        count += 1
print(f"At least 7 heads in {count} of 1000 runs, so roughly {count / 1000:.2f}")
````

**Watch for:** forgetting `return` (the function gives `None`); calling a function before running the cell that defines it; and mixing up `randint`, which includes its top number, with `range`, which does not.

### Ticket prices
A small theatre is setting the price of a ticket. About 200 people in town might come, and each of them has a most they are willing to pay – somewhere between £5 and £40, different for each person. Anyone whose limit is at or above the ticket price buys a ticket. A cheap ticket sells to almost everyone but earns little from each; a dear ticket earns a lot from each but sells to few. Which price brings in the most money?

`takings(price)` simulates 100 nights at a given price and returns the average takings per night. Notice it has a loop inside a loop: for each night, it goes through every person in town.

**Predict first.** Before you run anything, guess which price will win: £10, £20 or £30?

Run the cell below to define the function. Later cells rely on it.

````
import random

def takings(price, people=200, nights=100):
    """Average takings per night at a given ticket price, over many random nights."""
    total = 0
    for night in range(nights):
        for person in range(people):
            most = random.randint(5, 40)    # the most this person will pay, in £
            if most >= price:
                total += price              # they buy a ticket
    return round(total / nights, 2)
````

Run `takings` for every price from £5 to £40 in steps of £5, store the results, and plot takings against price. `range(5, 41, 5)` gives 5, 10, …, 40 – the third number is the step.

````
import matplotlib.pyplot as plt

random.seed(1)
prices = list(range(5, 41, 5))
results = []
for price in prices:
    results.append(takings(price))

best = prices[results.index(max(results))]
print(f"Best price: £{best}, average takings £{max(results):.2f} a night")

plt.figure()
plt.plot(prices, results, marker="o")
plt.xlabel("Ticket price (£)")
plt.ylabel("Average takings per night (£)")
plt.show()
````
